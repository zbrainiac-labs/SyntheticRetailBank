# FINMA LCR Monitoring Dashboard
## Liquidity Coverage Ratio - Real-Time Compliance Monitoring

**Regulatory Basis**: FINMA Circular 2015/2, LiqV Art. 14-20, Basel III LCR Framework  
**Reporting Frequency**: Daily (60-minute refresh)  
**Compliance Minimum**: 100% LCR  
**Internal Target**: 110% LCR  

---

## Overview

This dashboard provides **real-time monitoring** of the Liquidity Coverage Ratio (LCR) for compliance with Swiss regulatory requirements. The LCR measures a bank's ability to survive a 30-day stressed liquidity scenario.

### The LCR Formula

```
LCR = (High-Quality Liquid Assets / Net Cash Outflows over 30 days) × 100%
```

**Compliance Thresholds**:
- **≥ 110%**: PASS (above internal buffer target)
- **100% to < 110%**: WARNING (compliant, but below internal target)
- **< 100%**: FAIL (regulatory breach - notify FINMA, remediation plan)

**Calculation**: HQLA after L2B 15% cap and L2 40% cap; net outflows = stressed outflows (deposits, interbank funding, facility drawdowns) minus 30-day inflows capped at 75% of outflows. One shared formula (`REPP_AGG_FN_LCR_CALC`) drives the reported LCR, the stress scenarios and the what-if simulator.

## 6. Stress Scenarios

The engine evaluates every position under four scenarios simultaneously (`REPP_AGG_DT_LCR_SCENARIO_DAILY`):
**BASE** (regulatory FINMA LCR), **BANK_STRESS** (higher retail/corporate run-off, more facility drawdowns),
**MARKET_STRESS** (extra HQLA haircuts, lower inflows) and **COMBINED**. Rules live in `LIQI_RAW_TB_BEHAVIORAL_RULES`.

````python
# LCR per scenario over time + latest-day comparison
df_scen = session.sql("""
    SELECT AS_OF_DATE, SCENARIO_NAME, STRESS_SCENARIO, LCR_RATIO, LCR_STATUS, HQLA_TOTAL, NET_OUTFLOW
    FROM REPP_AGG_DT_LCR_SCENARIO_DAILY
    ORDER BY AS_OF_DATE
""").to_pandas()

st.header("Stress Scenarios")
st.line_chart(df_scen.pivot(index='AS_OF_DATE', columns='SCENARIO_NAME', values='LCR_RATIO'))

latest = df_scen[df_scen['AS_OF_DATE'] == df_scen['AS_OF_DATE'].max()]
cols = st.columns(len(latest))
for col, (_, row) in zip(cols, latest.iterrows()):
    col.metric(row['SCENARIO_NAME'], f"{row['LCR_RATIO']:.1f}%", row['LCR_STATUS'], delta_color='off')

fail_days = df_scen[df_scen['LCR_RATIO'] < 100].groupby('SCENARIO_NAME').size()
st.caption("Days below 100% per scenario: " + ", ".join(f"{k}: {v}" for k, v in fail_days.items()))
````

## 7. Root Cause of LCR Movements

For each liquidity event (`LIQI_RAW_TB_EVENTS`) the top client drivers come from `REPP_AGG_VW_LCR_VARIANCE_DRIVERS`.
`CLIENT_IMPACT_PP` is the funding-adjusted impact: a CHF 50M withdrawal shows its full negative effect because the cash
leaves the SNB reserves, even though the client's own stressed outflow falls. The exact per-line contributions
(`REPP_AGG_VW_LCR_VARIANCE_ATTRIBUTION`) sum to the reported LCR change.

````python
# Event timeline with the #1 client driver and LCR move on each event day
df_root = session.sql("""
    WITH ev AS (
        SELECT * FROM REP_RAW_V001.LIQI_RAW_TB_EVENTS
        QUALIFY ROW_NUMBER() OVER (PARTITION BY EVENT_ID ORDER BY INSERT_TIMESTAMP_UTC DESC) = 1
    )
    SELECT ev.EVENT_ID, ev.EVENT_DATE, ev.EVENT_TYPE, ev.DESCRIPTION,
           ROUND(d.LCR_DELTA_PP, 2)     AS LCR_CHANGE_PP,
           d.COUNTERPARTY_NAME          AS TOP_CLIENT,
           d.POSITION_ID                AS TOP_POSITION,
           ROUND(d.CLIENT_IMPACT_PP, 2) AS CLIENT_IMPACT_PP
    FROM ev
    LEFT JOIN REPP_AGG_VW_LCR_VARIANCE_DRIVERS d
           ON d.AS_OF_DATE = ev.EVENT_DATE AND d.COMPARISON_TYPE = 'DOD'
          AND d.STRESS_SCENARIO = 'BASE' AND d.CLIENT_RANK = 1
    ORDER BY ev.EVENT_DATE
""").to_pandas()

st.header("Root Cause of LCR Movements")
st.dataframe(df_root, hide_index=True)

# Exactness check: line contributions sum to the LCR change on every day
residual = session.sql("""
    SELECT MAX(ABS(s)) AS MAX_RESIDUAL_PP FROM (
        SELECT AS_OF_DATE, SUM(CONTRIBUTION_PP) - ANY_VALUE(LCR_DELTA_PP) AS s
        FROM REPP_AGG_VW_LCR_VARIANCE_ATTRIBUTION
        WHERE COMPARISON_TYPE = 'DOD' AND STRESS_SCENARIO = 'BASE'
        GROUP BY AS_OF_DATE)
""").to_pandas()['MAX_RESIDUAL_PP'].iloc[0]
st.caption(f"Attribution exactness: max residual {residual:.8f} pp across all days")
````

## 8. What-if Simulator

`REPP_AGG_SP_LCR_SIMULATE(as_of_date, params_json)` recalculates the LCR with the shared formula on the compact component
table, so each run takes well under a second. It returns a step-by-step waterfall and the extra corporate withdrawal that
would bring the LCR to exactly 100% (breakeven). The same logic backs the app Simulator tab and the agent tool `LCR_What_If`.

````python
import json

# Example shock: corporate non-operational run-off +50%, +10pp L2A haircut, 3% price shock, CHF 50M withdrawal
params = {
    "preset": "BASE",
    "corp_non_operational_mult": 1.5,
    "haircut_addon_l2a_pp": 10,
    "price_shock_pct": 3,
    "withdrawal_chf": 50_000_000,
}
df_sim = session.sql("CALL REPP_AGG_SP_LCR_SIMULATE(?, ?)", params=["", json.dumps(params)]).to_pandas()
df_sim.columns = [c.strip('"') for c in df_sim.columns]

st.header("What-if Simulator")
first, last = df_sim.iloc[0], df_sim.iloc[-1]
c1, c2, c3 = st.columns(3)
c1.metric("Baseline LCR", f"{first['LCR_RATIO']:.1f}%")
c2.metric("Simulated LCR", f"{last['LCR_RATIO']:.1f}%", f"{last['LCR_RATIO'] - first['LCR_RATIO']:+.1f}pp")
c3.metric("Extra withdrawal to 100%", f"CHF {last['BREAKEVEN_WITHDRAWAL_CHF'] / 1e6:,.0f}M")
st.dataframe(df_sim[['STEP_NAME', 'LCR_RATIO', 'DELTA_PP', 'LCR_STATUS']], hide_index=True)
````
